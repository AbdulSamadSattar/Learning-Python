In function definition:
def func(**kwargs):
➡️ Collect keyword arguments into a dictionary.

In function call:
func(**dictionary)
➡️ Unpack dictionary into keyword arguments.

*args
Collects positional arguments into a tuple:

**kwargs

Collects keyword arguments into a dictionary:

| Where you use `**` | Meaning                                  |
| ------------------ | ---------------------------------------- |
| `2 ** 3`           | Power                                    |
| `{**dict1}`        | Dictionary unpacking                     |
| `def f(**kwargs)`  | Collect keyword arguments                |
| `f(**dict1)`       | Unpack dictionary into keyword arguments |


```python

x = 2 ** 3
print(x)
```

```python
dict1 = {"name": "Ali"}
dict2 = {"age": 25}

result = {**dict1, **dict2}

print(result)
```

```python
{**dict1}
```

```python
a = {"x": 10, "y": 20}
b = {"z": 30}

c = {**a, **b}

print(c)
```

# ** in a function — Collect keyword arguments

```python
def student(**details):
    print(details)

student(name="Ali", age=25, city="Karachi")
```

** can also UNPACK a dictionary into a function

```python
def student(name, age):
    print(name, age)

data = {
    "name": "Ali",
    "age": 25
}

student(**data)
```

```python
def test(**kwargs):
    print(kwargs)

test(name="Ali", age=25)
```

```python
def test(*args):
    print(args)

test(10, 20, 30)
```
